```python
import pandas as pd
from sqlalchemy import create_engine, text
from IPython.display import display
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
import numpy as np
from datetime import datetime, timedelta
engine = create_engine('postgresql+psycopg2://airflow15:airflow15@localhost:5435/airflow15')

schema = 'pos_raw_retails_data'
```

```python
import pandas as pd
import numpy as np
from datetime import datetime, timedelta

# Number of rows you want to generate
n_rows = 10  

# Generate synthetic data
start_date = datetime(2025, 8, 1)
data = {
    "transaction_id": range(1, n_rows + 1),
    "customer_id": np.random.randint(100000, 100020, size=n_rows),
    "store_id": np.random.randint(1, 6, size=n_rows),
    "employee_id": np.random.randint(200, 220, size=n_rows),
    "transaction_date": [start_date + timedelta(days=i) for i in range(n_rows)],
    "total_amount": np.round(np.random.uniform(50, 1000, size=n_rows), 2),
    "payment_id": np.random.randint(300, 320, size=n_rows),
    "load_timestamp": [datetime.now()] * n_rows
}

df = pd.DataFrame(data)

# Display sample
display(df)

# Insert into Postgres
df.to_sql(
    "raw_sales_transactions",
    con=engine,
    schema=schema,
    if_exists="append",   # append data into table
    index=False
)

print(f"Inserted {len(df)} rows into {schema}.raw_sales_transactions")
```
